The session in our application stores the following information:

1. **User ID**:
   - session['user_id'] : Stores the CustomerID if the user is a customer or the `RestaurantID` if the user is a restaurant.

2. **Role**:
   - session['role'] : Stores the role of the user, either 'customer' or 'restaurant'.

3. **Customer Data** (if the user is a customer):
   - session['customer']: Stores a dictionary containing all the customer details from the `customers` table, such as:
     - `CustomerID`
     - `FirstName`
     - `LastName`
     - `Address`
     - `ZipCode`
     - `PhoneNumber`
     - `Password`
     - `Role`
     - `Balance`
     - `CreatedAt`

4. **Restaurant Data** (if the user is a restaurant):
   - session['restaurant']: Stores a dictionary containing all the restaurant details from the `restaurants` table, such as:
     - `RestaurantID`
     - `Name`
     - `Email`
     - `Address`
     - `ZipCode`
     - `Description`
     - `ImageURL`
     - `Password`
     - `Balance`
     - `OpenTime`
     - `CloseTime`
     - `CreatedAt`

5. **Shopping Cart** (if the user is a customer and has added items to the cart):
   - session['shoppingcart']: Stores a dictionary of items added to the cart, where each key is the ItemID and the value is another dictionary containing:
     - `Name`
     - `ItemToAdd`
     - `Quantity`
     - `Price`
     - `TotalPrice`

6. **Total Quantity** (if the user is a customer and has added items to the cart):
   - session['total_quantity']: Stores the total quantity of items in the shopping cart.

7. **Total Price** (if the user is a customer and has added items to the cart):
   - session['total_price']: Stores the total price of items in the shopping cart.

8. **Chosen Restaurant ID** (if the user is a customer and has selected a restaurant):
   - session['chosenrestID']: Stores the `RestaurantID` of the chosen restaurant.

This information is used throughout the application to manage user authentication, authorization, and the shopping cart functionality.
